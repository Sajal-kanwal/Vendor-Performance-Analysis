# Procurement–Sales Data Overview & Analytical Context

## 1. Data Landscape Overview

This dataset represents an end-to-end **procurement-to-sales pipeline**, capturing how products move from vendors through purchasing and logistics, and finally into sales.  
Each table represents a distinct operational layer, and when combined, they enable comprehensive **cost analysis, margin evaluation, and vendor performance assessment**.

---

## 2. Purchases Table — Procurement Transactions

The `purchases` table contains **actual procurement transaction data**, including:

- Purchase date  
- Vendor  
- Product brand  
- Quantity purchased  
- Total amount paid (USD)

This table reflects **real cash outflows** and forms the foundational layer for procurement analysis.

### Analytical Significance
- Identifies vendor-wise and brand-wise purchasing patterns  
- Enables temporal and seasonality analysis  
- Supports procurement volume and spending trend evaluation  
- Serves as the base for unit cost and efficiency calculations  

---

## 3. Purchase Prices Table — Pricing Reference Layer

The `purchase_prices` table provides **product-level pricing information**, including:

- Actual price  
- Purchase price  

Each **vendor–brand combination is unique**, ensuring reliable and unambiguous price mapping across datasets.

### Analytical Significance
- Acts as a price validation and normalization layer  
- Enables detection of pricing discrepancies across vendors  
- Supports cost benchmarking and unit economics analysis  
- Essential for accurate margin computation  

---

## 4. Vendor Invoice Table — Logistics & Aggregation Layer

The `vendor_invoice` table aggregates data from the `purchases` table at the **vendor–purchase order (PO) level**, and includes:

- Total quantity purchased  
- Total dollar amount  
- Freight cost  

This table maintains uniqueness based on **vendor and PO number**, making it the authoritative source for logistics-related expenses.

### Analytical Significance
- Enables accurate landed cost calculations (product cost + freight)  
- Supports evaluation of vendor logistics efficiency  
- Helps quantify freight as a proportion of total procurement cost  
- Reveals hidden costs not visible in raw purchase transactions  

---

## 5. Sales Table — Revenue Generation Layer

The `sales` table captures **actual sales transactions**, including:

- Product brand sold  
- Quantity sold  
- Selling price  
- Revenue generated  

This table represents **cash inflows** and reflects market demand and pricing effectiveness.

### Analytical Significance
- Enables brand-level revenue and demand analysis  
- Supports trend and performance evaluation over time  
- When combined with procurement data, enables margin and profitability analysis  
- Facilitates inventory turnover and sales efficiency insights  

---

## 6. Need for a Unified Summary Table

Since critical analytical information is distributed across multiple tables, isolated analysis would result in incomplete insights.  
To enable end-to-end evaluation, a consolidated **summary table** is required.

### Summary Table Requirements
The unified dataset should include:

- Purchase transactions by vendor and brand  
- Corresponding sales transaction data  
- Freight costs attributed to each vendor  
- Actual product prices from vendors  

---

## 7. Business & Analytical Impact

Creating a unified summary table enables:

- End-to-end profitability analysis (procurement → logistics → sales)  
- Vendor comparison based on:
  - Cost efficiency  
  - Freight burden  
  - Revenue contribution  
- Identification of:
  - High-margin and low-margin brands  
  - Vendors with hidden logistics inefficiencies  
- A strong foundation for:
  - Exploratory Data Analysis (EDA)  
  - KPI dashboards  
  - Predictive cost and revenue modeling  

---

## 8. Key Insight

> **Only by integrating procurement, pricing, logistics, and sales data can vendor profitability and operational efficiency be accurately assessed.**

```python
from db_connect import query_to_df

vendor_sales_summary = query_to_df("""
WITH FreightSummary AS (
    SELECT
        VendorNumber,
        SUM(Freight) AS TotalFreightCost
    FROM vendor_invoice
    GROUP BY VendorNumber
),

PurchaseSummary AS (
    SELECT
        p.VendorNumber,
        p.VendorName,
        p.Brand,
        p.Description,
        p.PurchasePrice,
        pp.PurchasePrice   AS ReferencePurchasePrice,   -- renamed for clarity
        pp.Volume,
        SUM(p.Quantity)       AS TotalPurchaseQuantity,
        SUM(p.Dollars)        AS TotalPurchaseDollars
    FROM purchases p
    LEFT JOIN purchase_prices pp
        ON p.Brand = pp.Brand
        AND p.VendorNumber = pp.VendorNumber          -- critical fix
    WHERE p.PurchasePrice > 0
      -- Optional: AND p.ReceivingDate >= '2024-01-01'   -- add if needed
    GROUP BY
        p.VendorNumber, p.VendorName, p.Brand, p.Description,
        p.PurchasePrice, pp.PurchasePrice, pp.Volume
),

SalesSummary AS (
    SELECT
        VendorNo            AS VendorNumber,    -- standardize name
        Brand,
        SUM(SalesQuantity)  AS TotalSalesQuantity,
        SUM(SalesDollars)   AS TotalSalesDollars,
        AVG(SalesPrice)     AS AvgSalesPrice,         -- changed from SUM - more meaningful
        SUM(ExciseTax)      AS TotalExciseTax
    FROM sales
    -- Optional: WHERE SalesDate >= '2024-01-01'
    GROUP BY VendorNo, Brand
)

SELECT
    ps.VendorNumber,
    ps.VendorName,
    ps.Brand,
    ps.Description,
    ps.PurchasePrice,
    ps.ReferencePurchasePrice,
    ps.Volume,
    ps.TotalPurchaseQuantity,
    ps.TotalPurchaseDollars,
    COALESCE(ss.TotalSalesQuantity, 0)      AS TotalSalesQuantity,
    COALESCE(ss.TotalSalesDollars, 0)       AS TotalSalesDollars,
    ss.AvgSalesPrice,
    COALESCE(ss.TotalExciseTax, 0)          AS TotalExciseTax,
    fs.TotalFreightCost                     AS VendorTotalFreight,

    -- Quick derived fields (expand in pandas if preferred)
    ROUND(
        COALESCE(ss.TotalSalesDollars, 0) - ps.TotalPurchaseDollars -
        COALESCE(fs.TotalFreightCost * (ps.TotalPurchaseDollars / NULLIF(SUM(ps.TotalPurchaseDollars) OVER (PARTITION BY ps.VendorNumber), 0)), 0)
    , 2) AS EstimatedGrossProfitProxy

FROM PurchaseSummary ps
LEFT JOIN SalesSummary ss
    ON ps.VendorNumber = ss.VendorNumber
   AND ps.Brand = ss.Brand
LEFT JOIN FreightSummary fs
    ON ps.VendorNumber = fs.VendorNumber

ORDER BY ps.TotalPurchaseDollars DESC
""")

# Quick inspection
print("Rows × Columns:", vendor_sales_summary.shape)
display(vendor_sales_summary.head(12))
```

```python
from db_connect import query_to_df
import matplotlib

vendor_brand_summary = query_to_df("""
WITH FreightSummary AS (
    SELECT
        VendorNumber,
        SUM(Freight) AS TotalFreightCost
    FROM vendor_invoice
    WHERE InvoiceDate >= '2024-01-01'
    GROUP BY VendorNumber
),

PurchaseSummary AS (
    SELECT
        p.VendorNumber,
        p.VendorName,
        p.Brand,
        -- Take the most frequent / latest description to reduce duplication
        ANY_VALUE(p.Description) AS MostCommonDescription,
        -- Use the weighted average purchase price as primary metric
        ROUND(SUM(p.Dollars) / NULLIF(SUM(p.Quantity), 0), 4) AS WeightedAvgPurchasePrice,
        ANY_VALUE(pp.PurchasePrice) AS ReferencePurchasePrice,
        ANY_VALUE(pp.Volume) AS VolumeML,
        SUM(p.Quantity) AS TotalPurchaseQty,
        SUM(p.Dollars) AS TotalPurchaseCost,
        COUNT(DISTINCT p.PONumber) AS DistinctPOs,
        MAX(p.ReceivingDate) AS LastReceiptDate
    FROM purchases p
    LEFT JOIN purchase_prices pp
        ON p.Brand = pp.Brand
        AND p.VendorNumber = pp.VendorNumber
    WHERE p.PurchasePrice > 0
      AND p.ReceivingDate >= '2024-01-01'
    GROUP BY
        p.VendorNumber, p.VendorName, p.Brand
),

SalesSummary AS (
    SELECT
        VendorNo AS VendorNumber,
        Brand,
        SUM(SalesQuantity) AS TotalSalesQty,
        SUM(SalesDollars) AS TotalSalesRevenue,
        ROUND(AVG(SalesPrice), 4) AS AvgSellingPrice,
        SUM(ExciseTax) AS TotalExciseTax,
        MAX(SalesDate) AS LastSaleDate,
        COUNT(DISTINCT SalesDate) AS SellingDays
    FROM sales
    WHERE SalesDate >= '2024-01-01'
    GROUP BY VendorNo, Brand
)

SELECT
    ps.VendorNumber,
    ps.VendorName,
    ps.Brand,
    ps.MostCommonDescription,
    ps.VolumeML,
    ps.WeightedAvgPurchasePrice,
    ps.ReferencePurchasePrice,
    ps.TotalPurchaseQty,
    ps.TotalPurchaseCost,
    ps.DistinctPOs,
    ps.LastReceiptDate,

    COALESCE(ss.TotalSalesQty, 0)          AS TotalSalesQty,
    COALESCE(ss.TotalSalesRevenue, 0)      AS TotalSalesRevenue,
    ss.AvgSellingPrice,
    COALESCE(ss.TotalExciseTax, 0)         AS TotalExciseTax,
    ss.LastSaleDate,
    ss.SellingDays,

    fs.TotalFreightCost                     AS VendorTotalFreight,

    -- Better freight allocation: proportional to purchase cost share within vendor
    ROUND(
        COALESCE(
            fs.TotalFreightCost *
            (ps.TotalPurchaseCost / NULLIF(SUM(ps.TotalPurchaseCost) OVER (PARTITION BY ps.VendorNumber), 0)),
            0
        ), 2
    ) AS AllocatedFreight,

    -- Landed cost = purchase cost + allocated freight
    ROUND(
        ps.TotalPurchaseCost +
        COALESCE(
            fs.TotalFreightCost *
            (ps.TotalPurchaseCost / NULLIF(SUM(ps.TotalPurchaseCost) OVER (PARTITION BY ps.VendorNumber), 0)),
            0
        ), 2
    ) AS TotalLandedCost,

    -- Core profitability metrics
    ROUND(
        COALESCE(ss.TotalSalesRevenue, 0) -
        (ps.TotalPurchaseCost +
         COALESCE(
             fs.TotalFreightCost *
             (ps.TotalPurchaseCost / NULLIF(SUM(ps.TotalPurchaseCost) OVER (PARTITION BY ps.VendorNumber), 0)),
             0
         ))
    , 2) AS EstimatedGrossProfit,

    ROUND(
        CASE
            WHEN COALESCE(ss.TotalSalesRevenue, 0) = 0 THEN NULL
            ELSE 100.0 * (
                COALESCE(ss.TotalSalesRevenue, 0) -
                (ps.TotalPurchaseCost +
                 COALESCE(
                     fs.TotalFreightCost *
                     (ps.TotalPurchaseCost / NULLIF(SUM(ps.TotalPurchaseCost) OVER (PARTITION BY ps.VendorNumber), 0)),
                     0
                ))
            ) / COALESCE(ss.TotalSalesRevenue, 0)
        END, 2
    ) AS GrossMarginPct,

    -- Velocity & performance indicators
    ROUND(
        CASE WHEN ps.TotalPurchaseQty = 0 THEN NULL
             ELSE ss.TotalSalesQty / ps.TotalPurchaseQty
        END, 3
    ) AS SellThroughRatio,

    -- Days since last activity (helps spot slow/dead items)
    DATEDIFF(CURDATE(), ps.LastReceiptDate) AS DaysSinceLastPurchase,
    DATEDIFF(CURDATE(), ss.LastSaleDate)    AS DaysSinceLastSale

FROM PurchaseSummary ps
LEFT JOIN SalesSummary ss
    ON ps.VendorNumber = ss.VendorNumber
   AND ps.Brand = ss.Brand
LEFT JOIN FreightSummary fs
    ON ps.VendorNumber = fs.VendorNumber

ORDER BY EstimatedGrossProfit DESC, TotalSalesRevenue DESC
""")

# ────────────────────────────────────────────────
# Quick post-processing & inspection in pandas
# ────────────────────────────────────────────────

print(f"Rows × Columns: {vendor_brand_summary.shape}")
print(f"Unique vendors: {vendor_brand_summary['VendorNumber'].nunique()}")
print(f"Unique brands:  {vendor_brand_summary['Brand'].nunique()}")

# Format numbers nicely for display
# styled = vendor_brand_summary.head(15).style.format({
#     'TotalPurchaseCost':   '${:,.2f}',
#     'TotalSalesRevenue':   '${:,.2f}',
#     'AllocatedFreight':    '${:,.2f}',
#     'TotalLandedCost':     '${:,.2f}',
#     'EstimatedGrossProfit':'${:,.2f}',
#     'GrossMarginPct':      '{:,.1f} %',
#     'SellThroughRatio':    '{:.2f}',
#     'WeightedAvgPurchasePrice': '${:,.4f}',
#     'AvgSellingPrice':     '${:,.2f}'
# }).background_gradient(
#     subset=['EstimatedGrossProfit', 'GrossMarginPct', 'SellThroughRatio'],
#     cmap='RdYlGn'
# )

display(vendor_brand_summary)
```

# Cleaning of Vendor Sales Summary

```python
import pandas as pd
# 1. Remove leading/trailing whitespaces from VendorName
vendor_sales_summary['VendorName'] = vendor_sales_summary['VendorName'].str.strip()

# 2. Convert Volume to numeric (float) – handle non-numeric values safely
vendor_sales_summary['Volume'] = pd.to_numeric(vendor_sales_summary['Volume'], errors='coerce')

#Fill missing ReferencePurchasePrice with PurchasePrice when available
vendor_sales_summary['ReferencePurchasePrice'] = vendor_sales_summary['ReferencePurchasePrice'].fillna(
    vendor_sales_summary['PurchasePrice']
)
# Volume Imputation
# ------------------------
# A. Brand-level median volume (most reliable)
brand_volume_median = vendor_sales_summary.groupby('Brand')['Volume'].median()

# B. Overall median as global fallback
global_volume_median = vendor_sales_summary['Volume'].median()

# Apply: first try brand median, then fall back to global
vendor_sales_summary['Volume'] = vendor_sales_summary.apply(
    lambda row: brand_volume_median.get(row['Brand'], global_volume_median)
    if pd.isna(row['Volume']) else row['Volume'],
    axis=1
)

#AvgSalesPrice Imputation
# ---------------------------------
# A. Brand-level median selling price (strongest signal)
brand_sales_median = vendor_sales_summary.groupby('Brand')['AvgSalesPrice'].median()

# B. Calculate typical markup per brand (where both prices exist)
valid_markup = vendor_sales_summary[
    (vendor_sales_summary['AvgSalesPrice'].notna()) &
    (vendor_sales_summary['ReferencePurchasePrice'].notna()) &
    (vendor_sales_summary['ReferencePurchasePrice'] > 0)
].copy()

valid_markup['ImpliedMarkup'] = (
    valid_markup['AvgSalesPrice'] / valid_markup['ReferencePurchasePrice']
)

brand_typical_markup = valid_markup.groupby('Brand')['ImpliedMarkup'].median()

# C. Global median markup as fallback
global_markup_median = valid_markup['ImpliedMarkup'].median() if not valid_markup.empty else 1.5

# Apply imputation logic
def impute_avg_sales_price(row):
    if pd.notna(row['AvgSalesPrice']):
        return row['AvgSalesPrice']

    # 1. Try brand median sales price
    brand_med = brand_sales_median.get(row['Brand'])
    if pd.notna(brand_med):
        return brand_med

    # 2. Try reference purchase price × typical brand markup
    ref_price = row['ReferencePurchasePrice']
    if pd.notna(ref_price) and ref_price > 0:
        markup = brand_typical_markup.get(row['Brand'], global_markup_median)
        return round(ref_price * markup, 2)

    # 3. Absolute fallback: overall median sales price
    return vendor_sales_summary['AvgSalesPrice'].median()

vendor_sales_summary['AvgSalesPrice_imputed'] = vendor_sales_summary.apply(
    impute_avg_sales_price, axis=1
)

vendor_sales_summary['GrossMarginPct'] = (
    (vendor_sales_summary['TotalSalesDollars'] - vendor_sales_summary['EstimatedGrossProfitProxy'])
    / vendor_sales_summary['TotalSalesDollars'].replace(0, pd.NA)
) * 100

vendor_sales_summary['SellThroughRatio'] = (
    vendor_sales_summary['TotalSalesQuantity'] / vendor_sales_summary['TotalPurchaseQuantity'].replace(0, pd.NA)
)
```

```python
import pandas as pd
import numpy as np

# ────────────────────────────────────────────────
# 1. Convert to numeric (fix object dtype)
# ────────────────────────────────────────────────
vendor_sales_summary['GrossMarginPct'] = pd.to_numeric(
    vendor_sales_summary['GrossMarginPct'],
    errors='coerce'
)

# ────────────────────────────────────────────────
# 2. Strong imputation logic for GrossMarginPct
# ────────────────────────────────────────────────

# Create mask for rows we need to impute
needs_imputation = vendor_sales_summary['GrossMarginPct'].isna()

# Option A – Most common & safest: set to 0% when no sales (no profit/loss realized)
vendor_sales_summary.loc[needs_imputation, 'GrossMarginPct'] = 0.0
```

```python
# 1. Simple gross profit (without freight)
vendor_sales_summary['GrossProfit'] = (
    vendor_sales_summary['TotalSalesDollars'] -
    vendor_sales_summary['TotalPurchaseDollars']
)

# 2. Profit margin – safe version (avoid division by zero)
vendor_sales_summary['ProfitMargin'] = np.where(
    vendor_sales_summary['TotalSalesDollars'] > 0,
    (vendor_sales_summary['GrossProfit'] / vendor_sales_summary['TotalSalesDollars']) * 100,
    0.0   # or np.nan if you prefer to keep missing
)

# 3. Stock turnover (sell-through ratio)
vendor_sales_summary['StockTurnover'] = np.where(
    vendor_sales_summary['TotalPurchaseQuantity'] > 0,
    vendor_sales_summary['TotalSalesQuantity'] / vendor_sales_summary['TotalPurchaseQuantity'],
    0.0   # or np.nan
)

# 4. Sales to Purchase ratio
vendor_sales_summary['SalesToPurchaseRatio'] = np.where(
    vendor_sales_summary['TotalPurchaseDollars'] > 0,
    vendor_sales_summary['TotalSalesDollars'] / vendor_sales_summary['TotalPurchaseDollars'],
    0.0   # or np.nan
)

# Quick check
print("New columns added:")
print(vendor_sales_summary[['GrossProfit', 'ProfitMargin', 'StockTurnover', 'SalesToPurchaseRatio']].describe())
```
